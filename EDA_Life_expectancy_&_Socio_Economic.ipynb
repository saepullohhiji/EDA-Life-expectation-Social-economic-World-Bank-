# Missing Values Analysis
Pada tahap ini, saya memeriksa apakah terdapat informasi yang tidak tersedia dalam dataset. Data yang kosong dapat memengaruhi hasil analisis, terutama ketika kita ingin membandingkan hubungan antarvariabel. Dengan mengetahui jumlah dan persentasenya, saya dapat menentukan strategi penanganan yang tepat tanpa menghilangkan terlalu banyak informasi atau menghasilkan kesimpulan yang bias

```python
missing_count = df.isnull().sum()
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100
```

```python
missing_table = pd.DataFrame({
    'Missing Values': missing_count,
    'Percentage (%)': missing_percentage
})
```

```python
missing_table = missing_table.sort_values(
    by='Percentage (%)',
    ascending=False
)

display(missing_table)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=missing_table.index,
    y=missing_table['Percentage (%)']
)

plt.title('Percentage of Missing Values')
plt.xlabel('Columns')
plt.ylabel('Missing Values (%)')
plt.xticks(rotation=75)
plt.tight_layout()
plt.show()
```

**Hasil Analisis**

Hasil pemeriksaan menunjukkan bahwa dataset masih memiliki beberapa informasi yang belum lengkap. Permasalahan terbesar ditemukan pada variabel Corruption, dengan sekitar 70,51% data tidak tersedia. Selain itu, variabel Sanitation dan Education Expenditure juga memiliki informasi kosong dalam jumlah cukup besar.

# Distribution Analysis
Pada tahap ini, saya memeriksa bagaimana nilai setiap variabel tersebar. Tujuannya adalah mengetahui apakah sebagian besar data berada pada rentang tertentu, apakah terdapat nilai yang terlalu tinggi atau rendah, serta apakah distribusi data memiliki karakteristik tertentu. Pemeriksaan ini penting karena nilai ekstrem dan distribusi yang tidak seimbang dapat memengaruhi hasil analisis statistik maupun proses pembangunan model Machine Learning.

```python
selected_columns = [
    'Life Expectancy World Bank',
    'Prevelance of Undernourishment',
    'Health Expenditure %',
    'Education Expenditure %',
    'Unemployment',
    'Sanitation'
]
```

```python
df[selected_columns].hist(
    figsize=(15, 10),
    bins=30,
    edgecolor='black'
)

plt.suptitle('Distribution Analysis', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 7))

sns.boxplot(data=df[selected_columns])

plt.title('Boxplot of Selected Variables')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
for column in selected_columns:

    plt.figure(figsize=(8, 4))

    sns.boxplot(x=df[column])

    plt.title(f'Boxplot of {column}')

    plt.tight_layout()
    plt.show()
```

```python
for column in selected_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\nVariabel: {column}")
    print("Lower Bound:", round(lower_bound, 2))
    print("Upper Bound:", round(upper_bound, 2))
    print("Jumlah Outlier:", len(outliers))
```

**Hasil Analisis**

Hasil analisis distribusi menunjukkan bahwa setiap indikator memiliki pola persebaran yang berbeda. Angka harapan hidup cenderung memiliki sejumlah nilai rendah yang membuat distribusinya miring ke kiri (Left-Skewed). Sebaliknya, indikator kekurangan gizi, pengeluaran kesehatan, pengeluaran pendidikan, dan pengangguran cenderung memiliki distribusi miring ke kanan (Right-Skewed).

Selain itu, pemeriksaan menggunakan metode IQR menemukan sejumlah nilai ekstrem, terutama pada variabel pengangguran dan kekurangan gizi.

# Correlation Analysis
Saya melakukan analisis korelasi untuk mengetahui faktor-faktor yang memiliki hubungan dengan angka harapan hidup. Dengan analisis ini, kita dapat melihat apakah suatu faktor cenderung meningkat bersama angka harapan hidup atau justru memiliki hubungan yang berlawanan. Hasilnya dapat membantu mengidentifikasi variabel yang menarik untuk dianalisis lebih lanjut. Namun, hubungan antarvariabel tidak selalu berarti bahwa satu variabel menyebabkan perubahan pada variabel lainnya.

```python
numeric_df = df.select_dtypes(include=np.number)
```

```python
correlation_matrix = numeric_df.corr(
    method='pearson'
)
```

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```

```python
target_correlation = correlation_matrix[
    'Life Expectancy World Bank'
].drop('Life Expectancy World Bank')
```

```python
target_correlation = target_correlation.sort_values(
    ascending=False
)
```

```python
display(target_correlation)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=target_correlation.values,
    y=target_correlation.index
)

plt.title('Correlation with Life Expectancy')
plt.xlabel('Correlation Coefficient')
plt.ylabel('Variables')
plt.axvline(x=0, color='black', linewidth=1)

plt.tight_layout()
plt.show()
```

**Hasil Analisis**

Berdasarkan analisis korelasi, ditemukan bahwa sanitasi dan prevalensi kekurangan gizi memiliki hubungan yang relatif kuat dengan angka harapan hidup.

Sanitasi menunjukkan hubungan positif sebesar 0,680. Artinya, negara dengan tingkat sanitasi yang lebih baik cenderung memiliki angka harapan hidup lebih tinggi dalam data yang dianalisis.

# Pattern Analysis
Pada tahap ini, saya mencari pola dalam data berdasarkan waktu, wilayah, dan kelompok pendapatan negara. Tujuannya adalah mengetahui apakah terdapat perbedaan atau kecenderungan tertentu dalam angka harapan hidup. Hasil analisis ini dapat memberikan gambaran yang lebih mendalam mengenai kondisi setiap kelompok dan membantu menentukan aspek mana yang perlu diteliti lebih lanjut.

**A. Menganalisis pola berdasarkan tahun**

```python
yearly_pattern = df.groupby('Year')[
    'Life Expectancy World Bank'
].mean()
```

```python
display(yearly_pattern)
```

```python
plt.figure(figsize=(12, 5))

sns.lineplot(
    x=yearly_pattern.index,
    y=yearly_pattern.values,
    marker='o'
)

plt.title('Life Expectancy Trend (2001-2019)')
plt.xlabel('Year')
plt.ylabel('Average Life Expectancy')

plt.grid(True)
plt.tight_layout()
plt.show()
```

**Hasil Analisis**

Berdasarkan analisis perkembangan dari tahun ke tahun, ditemukan bahwa rata-rata angka harapan hidup mengalami peningkatan secara konsisten selama periode 2001 hingga 2019. Pada awal periode, rata-rata angka harapan hidup berada di sekitar 66,67 tahun dan meningkat menjadi 72,58 tahun pada akhir periode.

**B. Menganalisis pola berdasarkan wilayah**

```python
region_pattern = df.groupby('Region')[
    'Life Expectancy World Bank'
].mean().sort_values(ascending=False)
```

```python
display(region_pattern)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=region_pattern.values,
    y=region_pattern.index
)

plt.title('Average Life Expectancy by Region')
plt.xlabel('Average Life Expectancy')
plt.ylabel('Region')

plt.tight_layout()
plt.show()
```

**Hasil Analisis**

Hasil analisis berdasarkan wilayah menunjukkan adanya perbedaan angka harapan hidup yang cukup besar. North America memiliki rata-rata tertinggi, yaitu sekitar 79,68 tahun, sedangkan Sub-Saharan Africa memiliki rata-rata terendah sebesar 57,54 tahun.

**C. Menganalisis pola berdasarkan kelompok pendapatan**

```python
income_pattern = df.groupby('IncomeGroup')[
    'Life Expectancy World Bank'
].mean().sort_values(ascending=False)
```

```python
display(income_pattern)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=income_pattern.index,
    y=income_pattern.values
)

plt.title('Average Life Expectancy by Income Group')
plt.xlabel('Income Group')
plt.ylabel('Average Life Expectancy')

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

**Hasil Analisis**

Hasil analisis berdasarkan kelompok pendapatan menunjukkan adanya pola yang cukup jelas. Negara-negara dengan kelompok pendapatan tinggi memiliki rata-rata angka harapan hidup sebesar 78,31 tahun, sedangkan negara dengan kelompok pendapatan rendah memiliki rata-rata sebesar 56,47 tahun.

Perbedaan tersebut menunjukkan bahwa terdapat kesenjangan angka harapan hidup berdasarkan kelompok pendapatan negara.
